```python
import pandas as pd
import shap
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report
#Reading the files
customers = pd.read_csv(r"/content/drive/MyDrive/olist_customers_dataset.csv")
geolocation = pd.read_csv(r"/content/drive/MyDrive/olist_geolocation_dataset.csv")
order_items = pd.read_csv(r"/content/drive/MyDrive/olist_order_items_dataset.csv")
order_payments = pd.read_csv(r"/content/drive/MyDrive/olist_order_payments_dataset.csv")
order_reviews = pd.read_csv(r"/content/drive/MyDrive/olist_order_reviews_dataset.csv")
olist_orders = pd.read_csv(r"/content/drive/MyDrive/olist_orders_dataset.csv")
products = pd.read_csv(r"/content/drive/MyDrive/olist_products_dataset.csv")
sellers = pd.read_csv(r"/content/drive/MyDrive/olist_sellers_dataset.csv")
translation = pd.read_csv(r"/content/drive/MyDrive/product_category_name_translation.csv")
# Merging items, products and translation
items_products = pd.merge(order_items,products, on = "product_id", how = "inner")
items_products = pd.merge(items_products,sellers, on = "seller_id", how = "left")
items_products = pd.merge(items_products,translation, on = "product_category_name", how = "left")
# Grouping all order_ids in order to group orders by their products.
items_agg = items_products.groupby('order_id').agg({'price':'sum', 'freight_value':'sum', 'product_id':'count', 'seller_id':'nunique', 'product_category_name':'nunique', 'product_category_name_english': 'first'}).rename(columns={'product_id': 'n_items', 'seller_id': 'n_sellers', 'product_category_name': 'n_categories'})
merged_df = pd.merge(olist_orders,items_agg, on = "order_id", how = "inner")
merged_df = pd.merge(merged_df,order_reviews, on = "order_id", how = "inner")
# Merging databases
merged_df = pd.merge(merged_df,order_payments, on = "order_id", how = "left")
merged_df = pd.merge(merged_df,customers, on = "customer_id", how = "left")
merged_df = merged_df.drop_duplicates(subset='order_id', keep = "first")
location_agg = geolocation.groupby('geolocation_zip_code_prefix').agg({'geolocation_lng':'mean','geolocation_lat':'mean'}).rename(columns = {'geolocation_lng':'geolocation_lng_mean','geolocation_lat':'geolocation_lat_mean'}).reset_index()
merged_df = pd.merge(merged_df,location_agg, left_on="customer_zip_code_prefix", right_on="geolocation_zip_code_prefix", how = "left")
# Finally after getting all databases connected we can add target section
merged_df['target'] =(merged_df['review_score']<=3).astype(int)

delivery_days = pd.to_datetime(merged_df['order_delivered_customer_date'])
estimated_days = pd.to_datetime(merged_df['order_estimated_delivery_date'])
order_purchase_timestamp = pd.to_datetime(merged_df['order_purchase_timestamp'])
order_delivery_carrier_timestamp = pd.to_datetime(merged_df['order_delivered_carrier_date'])
merged_df['time before delivery'] = (order_delivery_carrier_timestamp - order_purchase_timestamp).dt.days
merged_df['time during delivery'] = (delivery_days- order_delivery_carrier_timestamp).dt.days
merged_df['delivery_days'] = (delivery_days - estimated_days).dt.days
filtered_df = merged_df.groupby('target')[['price', 'freight_value', 'delivery_days','payment_installments', 'time before delivery']].mean().rename(columns = {'price':'price_mean', 'freight_value':'freight_value_mean'})
filtered_df = filtered_df.reset_index()
#print(merged_df.columns.tolist())
filtered_df
merged_df = merged_df.drop(merged_df[((merged_df.order_status == 'canceled') | (merged_df.order_status ==  'unavailable'))].index)
merged_df = merged_df.sort_values(by = 'order_purchase_timestamp')
merged_df = merged_df.drop(columns = [
    'order_id','n_categories','customer_id', 'review_comment_message',
    'review_comment_title','order_purchase_timestamp','order_approved_at',
    'order_delivered_carrier_date','order_delivered_customer_date','order_estimated_delivery_date',
    'customer_unique_id', 'geolocation_zip_code_prefix'])
print(len(merged_df.product_category_name_english.unique()))
featured_cols = [
    'price', 'freight_value', 'n_items', 'n_sellers',
    'payment_sequential', 'payment_installments',
    'payment_value', 'payment_type',
    'customer_state','target', 'time before delivery', 'delivery_days', 'time during delivery'
]
model_df = merged_df[featured_cols].copy()
model_df = pd.get_dummies(model_df, columns = ['payment_type', 'customer_state'])
train, test = train_test_split(model_df, test_size=0.2, shuffle=False)
train = train.dropna()
test = test.dropna()
X_train = train.drop(columns=['target'])
y_train = train['target'].dropna()
X_test = test.drop(columns=['target'])
y_test = test['target']
#print(X_train.isnull().sum())
#print(X_test.isnull().sum())
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
clf = LogisticRegression(random_state=0, class_weight='balanced').fit(X_train_scaled, y_train)
y_pred = clf.predict(X_test_scaled)
print("X shape:", X_train_scaled.shape)
print("y shape:", y_train.shape)
#print(roc_auc_score(y_test, y_pred))
#print(confusion_matrix(y_test, y_pred))
#print(classification_report(y_test, y_pred))
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
scale_weight = neg_count / pos_count

models = {
    'LogReg': LogisticRegression(random_state=0),
    'LogReg balanced': LogisticRegression(random_state=0, class_weight='balanced'),
    'XGBoost': XGBClassifier(random_state=0),
    'XGBoost + scale_pos_weight': XGBClassifier(random_state=0, scale_pos_weight=scale_weight),
}
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    proba = model.predict_proba(X_test_scaled)[:, 1]
    print(f"{name}: ROC-AUC = {roc_auc_score(y_test, proba):.3f}")

clf_xgb = models['XGBoost + scale_pos_weight']
y_pred_xgb = clf_xgb.predict(X_test_scaled)
y_proba_xgb = clf_xgb.predict_proba(X_test_scaled)[:, 1]
print(roc_auc_score(y_test, y_proba_xgb))
print(confusion_matrix(y_test, y_pred_xgb))
print(classification_report(y_test, y_pred_xgb))
print(merged_df[['price', 'freight_value', 'delivery_days', 'time before delivery', 'target', 'n_items', 'n_sellers', 'payment_installments', 'time during delivery']].corr()['target'])
explainer = shap.TreeExplainer(clf_xgb)
shap_values = explainer.shap_values(X_test_scaled)
shap.summary_plot(shap_values, X_test_scaled, feature_names=X_test.columns)
import numpy as np
threshold = np.quantile(y_proba_xgb, 0.8)
is_top20 = y_proba_xgb >= threshold
caught = y_test[is_top20].sum()
caught_pct = caught / y_test.sum()
print(f"threshold={threshold:.4f}, caught={caught}, share={caught_pct:.1%}, lift={caught_pct / 0.2:.2f}x")
```
